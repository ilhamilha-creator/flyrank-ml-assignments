## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**The Decision:** Which pages should content teams prioritize for refresh/update review?

**The Actor:** Content editors, SEO specialists, and content strategists who decide where to invest editorial time.

**Cost of a Wrong Call:** A False Positive (flagging a healthy page for refresh) wastes valuable editorial time that could have been spent on pages that actually need attention. A False Negative (missing a declining page) means lost organic traffic and revenue that could have been recovered with timely content updates.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Define realistic cost structure for content refresh decisions
editorial_hours_wasted = 4.0  # hours spent on unnecessary refresh
hourly_editorial_cost = 75.0  # average hourly cost for content work
cost_of_false_positive = editorial_hours_wasted * hourly_editorial_cost

lost_traffic_value = 2000.0  # estimated monthly value of organic traffic
recovery_rate = 0.3  # 30% of declining pages can be recovered with timely refresh
cost_of_false_negative = lost_traffic_value * recovery_rate

print(f"Cost of False Positive (Wasted Editorial Time): ${cost_of_false_positive:.2f}")
print(f"Cost of False Negative (Lost Traffic Value): ${cost_of_false_negative:.2f}")
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Based on the starter dataset (content_refresh_anonymized.csv), the analysis shows strong potential for this lane. The dataset contains 30,000 content pages with rich search performance signals. The baseline declining rate is 54.2%, indicating a substantial pool of pages that may need attention. The starter pipeline already demonstrated that a learned model can achieve 3.1x better Precision@50 than hand-written rules (0.740 vs 0.240), proving there's measurable signal to be captured for content refresh prioritization.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
import os

# Load the actual starter dataset
df = pd.read_csv('data/raw/content_refresh_anonymized.csv')

# Calculate real numbers from the data
total_pages = len(df)
declining_rate = (df["trend_direction"].str.lower() == "down").mean()
impressions_avg = df["impressions_90d"].mean()

print(f"Real Number 1 (Total Pages): {total_pages:,}")
print(f"Real Number 2 (Declining Rate): {declining_rate:.1%}")
print(f"Real Number 3 (Avg 90d Impressions): {impressions_avg:.0f}")

# Load model results from the starter pipeline
import json
with open('outputs/model_results.json', 'r') as f:
    model_results = json.load(f)

baseline_p50 = model_results["baseline"]["baseline_precision_at_50"]
rf_p50 = model_results["models"]["random_forest"]["precision_at_50"]
improvement = rf_p50 / baseline_p50

print(f"\nModel Performance Evidence:")
print(f"Baseline Precision@50: {baseline_p50:.3f}")
print(f"Random Forest Precision@50: {rf_p50:.3f}")
print(f"Model Improvement Factor: {improvement:.1f}x")
```
