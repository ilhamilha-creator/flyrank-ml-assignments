## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**The Decision:** Which pages should content teams prioritize for refresh/update review?

**The Actor:** Content editors, SEO specialists, and content strategists who decide where to invest editorial time.

**Cost of a Wrong Call:** A False Positive (flagging a healthy page for refresh) wastes valuable editorial time that could have been spent on pages that actually need attention. A False Negative (missing a declining page) means lost organic traffic and revenue that could have been recovered with timely content updates.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Define realistic cost structure for content refresh decisions
editorial_hours_wasted = 4.0  # hours spent on unnecessary refresh
hourly_editorial_cost = 75.0  # average hourly cost for content work
cost_of_false_positive = editorial_hours_wasted * hourly_editorial_cost

lost_traffic_value = 2000.0  # estimated monthly value of organic traffic
recovery_rate = 0.3  # 30% of declining pages can be recovered with timely refresh
cost_of_false_negative = lost_traffic_value * recovery_rate

print(f"Cost of False Positive (Wasted Editorial Time): ${cost_of_false_positive:.2f}")
print(f"Cost of False Negative (Lost Traffic Value): ${cost_of_false_negative:.2f}")
```
